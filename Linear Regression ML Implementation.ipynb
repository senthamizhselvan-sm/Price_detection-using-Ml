```python
from sklearn.utils import Bunch
import pandas as pd
import numpy as np

data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep=r"\s+", skiprows=22, header=None)

data = np.hstack([
    raw_df.values[::2, :],
    raw_df.values[1::2, :2]
])

target = raw_df.values[1::2, 2]

feature_names = np.array([
    "CRIM", "ZN", "INDUS", "CHAS", "NOX",
    "RM", "AGE", "DIS", "RAD", "TAX",
    "PTRATIO", "B", "LSTAT"
])

descr = """
.. _boston_dataset:

Boston house prices dataset
---------------------------

Data Set Characteristics:

:Number of Instances: 506

:Number of Attributes: 13 numeric/categorical predictive.

:Attribute Information (in order):
    - CRIM     per capita crime rate by town
    - ZN       proportion of residential land zoned
    - INDUS    proportion of non-retail business acres
    - CHAS     Charles River dummy variable
    - NOX      nitric oxides concentration
    - RM       average number of rooms per dwelling
    - AGE      proportion of owner-occupied units built before 1940
    - DIS      weighted distances to employment centres
    - RAD      accessibility to radial highways
    - TAX      property-tax rate
    - PTRATIO  pupil-teacher ratio
    - B        proportion of blacks by town
    - LSTAT    lower status of the population

:Target:
    MEDV - Median value of owner-occupied homes
"""

boston = Bunch(
    data=data,
    target=target,
    feature_names=feature_names,
    DESCR=descr
)

```

```python
### Independent and dependent features

X = dataset.iloc[: , : -1]
Y = dataset.iloc[: , -1]
```

```python
## train test split
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test , Y_train , Y_test = train_test_split(X , Y , test_size = 0.3 , random_state = 42)
```

```python
### standardize the dataset
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
X_train
```

### Model Training 

```python
from sklearn.linear_model import LinearRegression
```

```python
regression  = LinearRegression()
```

```python
regression.fit(X_train , Y_train)
```

```python
## print the coeffiens and the intercept
```

```python
print(regression.coef_)
```

```python
print(regression.intercept_)
```

```python
## on which argument the model has to be trained
regression.get_params()
```

```python
## prediction with test data
```

```python
reg_pred = regression.predict(X_test)
```

```python
reg_pred
```

```python
### plot a sctter plot for the prediction
```

```python
plt.scatter(Y_test , reg_pred)
plt.show()
```

```python
## residuals
```

```python
residuals = Y_test-reg_pred
```

```python
residuals
```

```python
## plot the residuals
```

```python
sns.displot(residuals , kind = "kde")
plt.show()
```

```python
## scatter plot with respect to prediction and residuals
```

```python
plt.scatter(reg_pred , residuals)
plt.show()
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
```

```python
print(mean_squared_error(Y_test , reg_pred))
print(mean_absolute_error(Y_test , reg_pred))
print(np.sqrt(mean_squared_error(Y_test , reg_pred)))
```

### R square and adjusted R square

```python
# FORMULA 
# r^2 = 1 - SSR / SST
# R^2 = coefficient of determination SSr = sum of squares of residuals 
# SST = total sum of squares
```

```python
from sklearn.metrics import r2_score
score = r2_score(Y_test , reg_pred)
```

```python
print(score)
```

```python
# # ADjusted R ^ 2 = 1-[(1-r2)*(n-1)/(n-k-1)]

# where :
# R2 : the R2 of thw model n : he number of observations K : the number of predictor values
```

```python
## display the adjusted r sqquare
```

```python
1 - (1-score)*(len(Y_test)-1) / (len(Y_test) - X_test.shape[1]-1)
```
